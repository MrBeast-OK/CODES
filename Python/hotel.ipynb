```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('clean_final_data.csv')

df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

plt.hist(df['Age'].dropna(), bins=10, edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Age Distribution')
plt.show()
```
