# Datasets In Seaborn

```python
sns.get_dataset_names()
iris = sns.load_dataset("iris")
titanic = sns.load_dataset("titanic")
penguins = sns.load_dataset("penguins")
flights = sns.load_dataset("flights")
tips = sns.load_dataset("tips")
```

# Seaborn Basic Plot Structure

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex")
```

```python
sns.countplot(
    data = iris,
    x = "species"
)
plt.show()
```

```python
sns.countplot(
    data = tips,
    x = "smoker",
    hue = "time"
)
plt.show()
```

```python
sns.histplot(
    data = tips,
    x = "total_bill",
    kde=True
)
plt.show()
```

HEATMAP

```python
correlation = titanic.select_dtypes("number").corr()

print(correlation)
```

Regression Plot

```python
sns.regplot(
    data = titanic,
    x = "age",
    y = "fare"
)
plt.show()
```

```python
titanic.isnull()
```

```python
sns.heatmap(
    titanic.isnull(),
    cbar=False
)
plt.show()
```
