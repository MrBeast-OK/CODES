```python
import pandas as pd

df = pd.read_csv("Airline Dataset Updated.csv")

print("--- Gender Count ---")
print(df['Gender'].value_counts())

print("\n--- Gender Percentage Distribution (%) ---")
print(df['Gender'].value_counts(normalize=True) * 100)

print("\n--- Top 10 Nationalities ---")
print(df['Nationality'].value_counts().head(10))

print("\n--- Passenger Count by Continent ---")
print(df['Continents'].value_counts())

print("\n--- Passenger Count by Country ---")
print(df['Country Name'].value_counts())

print("\n--- Top 15 Airport Countries ---")
print(df['Country Name'].value_counts().head(15))

print("\n--- Flight Status Count ---")
print(df['Flight Status'].value_counts())

print("\n--- Flight Status Percentage Distribution (%) ---")
print(df['Flight Status'].value_counts(normalize=True) * 100)

print("\n--- Unique Airport Names ---")
print(df['Airport Name'].nunique())
```
